# QFI Infrascope Data Analysis Example + Boilerplate Code

```python
from qfi_analyzer import QFIAnalyzer

analyzer = QFIAnalyzer(
    file_path='./example_data.txt',
    verbose=True
)
```

```python
# Generate a GIF of the QFI data
analyzer.generate_gif(
    save_to='./',
    file_name='example_gif.gif',
    fps=2,
)
```

```python
analyzer.generate_image(
    save_to='./',
    file_name='example_image.png',
)
```

```python
# 0, 0 starts at top left corner of the frame
pixel_value_1 = analyzer.get_pixel_value(frame=0,height=0,width=0)
pixel_value_2 = analyzer.get_pixel_value(frame=2,height=242,width=320)
pixel_value_3 = analyzer.get_pixel_value(frame=1,height=245,width=520)

# if you need faster array access you can use array indexing directly
pixel_value_1 = analyzer.pixel_array[0,0,0] # [frame, height, width]

print("pixel_value_1: ", pixel_value_1)
print("pixel_value_2: ", pixel_value_2)
print("pixel_value_3: ", pixel_value_3)
```

```python
frame_1 = analyzer.get_frame_array(frame=1)
print("frame_1: ", frame_1)
```

```python
maximum_value_in_frame_1 = analyzer.maximum_value_in_frame(frame=1)
minimum_value_in_frame_1 = analyzer.minimum_value_in_frame(frame=1)
average_value_in_frame_1 = analyzer.average_value_in_frame(frame=1)

print("maximum_value_in_frame_1: ", maximum_value_in_frame_1)
print("minimum_value_in_frame_1: ", minimum_value_in_frame_1)
print("average_value_in_frame_1: ", average_value_in_frame_1)
```
